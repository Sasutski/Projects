# A brief history about PI
Pi (*π*) is the mathematical constant representing the ratio of a circle's circumference to its diameter. It is often approximated as 3.14159, and it's history dates back roughly 4000 years of history, evolving from practical estimations to precise formulas.

## The Babylonians
The first ever approximation of Pi came from the Babylonians, around 1900 BC, whom approximated Pi by calculating the area of a circle by taking 3 times the square of its radius, which gave the value of Pi as 3. A closer approximation from the Babylonians was 3.125.
